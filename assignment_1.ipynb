WHO Global Health Observatory Dataset Features

- **GHO (CODE):** The identification code assigned by the World Health Organization to each health indicator.
- **GHO (DISPLAY):** The name or description of the health indicator being measured.
- **GHO (URL):** A link to the WHO webpage containing information about the indicator.
- **YEAR (DISPLAY):** The year associated with the reported result.
- **STARTYEAR:** The first year of the measurement period.
- **ENDYEAR:** The last year of the measurement period.
- **REGION (CODE):** The abbreviated WHO code for the geographic region. For example, `AFR` represents Africa.
- **REGION (DISPLAY):** The complete name of the WHO geographic region.
- **COUNTRY (CODE):** The three-letter code used to identify the country. For example, `ZMB` represents Zambia.
- **COUNTRY (DISPLAY):** The complete name of the country.
- **DIMENSION (TYPE):** The type of additional category used to separate the data, when applicable.
- **DIMENSION (CODE):** The code assigned to an additional data category, when applicable.
- **DIMENSION (NAME):** The name of the additional data category, when applicable.
- **Numeric:** The complete numerical result for the health indicator.
- **Value:** The displayed or rounded version of the numerical result.
- **Low:** The lower limit of the estimated result, when available.
- **High:** The upper limit of the estimated result, when available.

**Note:** `NaN` means that a value is missing or was not provided. The numbers shown on the far left are row indexes and are not dataset features.
